```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
pd.read_csv("Medical student Csv.csv")
```

```python
#saving the Dataframe to a variable, df
df=pd.read_csv("Medical student Csv.csv")
```

```python
#making a copy of the variable incase of a data lose
Data=df.copy()
```

```python
Q1=df.groupby("Satisfaction with health")["age"].quantile(0.25)
Q3=df.groupby("Satisfaction with health")["age"].quantile(0.75)
IQR=Q3-Q1
outliers_age = df[
    (df["age"] < (Q1[df['Satisfaction with health']].values - 1.5 * IQR[df['Satisfaction with health']].values)) |
    (df["age"] > (Q3[df['Satisfaction with health']].values + 1.5 * IQR[df['Satisfaction with health']].values))
]
outliers_age[["Satisfaction with health","age"]]
```

```python
n0_outlier_age= df[
    (df["age"] >(Q1[df['Satisfaction with health']].values - 1.5 * IQR[df['Satisfaction with health']].values)) &
    (df["age"] <(Q3[df['Satisfaction with health']].values + 1.5 * IQR[df['Satisfaction with health']].values))
]
n0_outlier_age
```

```python
df=n0_outlier_age
df
```
