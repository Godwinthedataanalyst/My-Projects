```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
from datetime import datetime
Healthcare=pd.read_csv("healthcare_dataset.csv")
Healthcare
```

```python
Healthcare.info()
```

```python
Healthcare["Date of Admission"]=pd.to_datetime(Healthcare["Date of Admission"])
Healthcare["Discharge Date"]=pd.to_datetime(Healthcare["Discharge Date"])
Healthcare.dtypes
```

```python
 Healthcare["Discharge Date"].dt.year.nunique()
```

```python
Healthcare["Month Name"]=Healthcare['Date of Admission'].dt.month_name()
Healthcare
```

```python
Healthcare["Admission Year"]=Healthcare["Date of Admission"].dt.year
Healthcare
```

```python
#showing groupby of "Medical Condition",["Admission Type"]==Emergency
Medical_Emergency=Healthcare[Healthcare['Admission Type']=="Emergency"].groupby('Medical Condition')['Admission Type'].value_counts()
Medical_Emergency
```

```python
Healthcare_2019=Healthcare[Healthcare['Date of Admission'].dt.year==2019]
```

```python
#showing groupby of "Medical Condition",["Admission Type"]==Emergency
Medical_19_Emergency=Healthcare_2019[Healthcare_2019['Admission Type']=="Emergency"].groupby('Medical Condition')['Admission Type'].value_counts()
Medical_19_Emergency
```

```python
Medical_19_Emergency.plot(kind="pie",autopct="%1.1f%%",explode=[0,0,0,0,0,0.1],shadow=True,figsize=(6,5))
plt.title("Medical Condition Emergency Admission")
plt.show()
```

```python
#Medical Condition and Billing Amount
Healthcare_Billing=Healthcare.groupby('Medical Condition')["Billing Amount"].sum()
Healthcare_Billing
```

```python
pd.options.display.float_format = '{:.2f}'.format
Healthcare_Billing
```

```python
#using a Barh to show the visualization
#using a Barh to show the visualization
Healthcare_Billing.plot(kind="bar",figsize=(8,3))
plt.xlabel("Billing Amount")
plt.xticks(rotation=45)
plt.title("Medical Condition Billing Amount")
plt.show()
```

```python
#Medical Condition and Billing Amount
Healthcare_Billing_Year=Healthcare.groupby('Admission Year')["Billing Amount"].sum()
Healthcare_Billing_Year
```

```python
#using a Barh to show the visualization
#using a Barh to show the visualization
Healthcare_Billing_Year.plot(kind="barh",figsize=(8,3),color="violet")
plt.xlabel("Billing Amount")
plt.xticks(rotation=45)
plt.title("Medical Condition Billing Amount")
plt.show()
```

```python
Highest_Year=Healthcare["Admission Year"].value_counts().sort_index(ascending=False)
Highest_Year
```

```python
#using a barh chart to visualize and putting the information on the Columns
Highest_Year.plot(kind="barh", figsize=(8,4),color="orange")
plt.ylabel=("Admission Year")
plt.xlabel("count")
plt.title("Year of Admission")
plt.show()
```
